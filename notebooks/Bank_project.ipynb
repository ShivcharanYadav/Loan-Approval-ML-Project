```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings(action="ignore")
```

```python
df=pd.read_csv("C:/Users/SHIVCHARAN/Documents/Project/project2/Loan_Approval_AI_Project/data/loan_data.csv",sep=",")
```

```python
df
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
df["person_gender"]=le.fit_transform(df["person_gender"])
```

```python
df["person_education"]=le.fit_transform(df["person_education"])
```

```python
df["person_home_ownership"]=le.fit_transform(df["person_home_ownership"])
```

```python
df["loan_intent"]=le.fit_transform(df["loan_intent"])
```

```python
df["previous_loan_defaults_on_file"]=le.fit_transform(df["previous_loan_defaults_on_file"])
```

```python
x=df.drop("loan_status",axis=1)
```

```python
y=df["loan_status"]
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.3,stratify=y)
```

```python
from sklearn.ensemble import AdaBoostClassifier,GradientBoostingClassifier
```

---

AdaBoostClassifier

```python
ada=AdaBoostClassifier()
```

```python
ada.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
y_pred=ada.predict(x_test)
```

```python
ada_accu=accuracy_score(y_test,y_pred)
```

```python
ada_accu
```

---

GradientBoostingClassifier

```python
gb=GradientBoostingClassifier()
```

```python
gb.fit(x_train,y_train)
```

```python
y_pred2=gb.predict(x_test)
```

```python
gb_accu=accuracy_score(y_test,y_pred2)
```

```python
gb_accu
```

---

```python
from xgboost import XGBClassifier
```

```python
xgb=XGBClassifier(n_estimators=100,learning_rate=0.1,use_label_encoder=False,eval_metric='mlogloss',random_state=42)
```

```python
xgb.fit(x_train,y_train)
```

```python
y_pred3=xgb.predict(x_test)
```

```python
xgb_accu=accuracy_score(y_test,y_pred3)
```

```python
xgb_accu
```
